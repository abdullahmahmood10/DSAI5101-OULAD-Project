```python
import time

from catboost import CatBoostClassifier
from xgboost import XGBClassifier

from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import (
    HistGradientBoostingClassifier,
    RandomForestClassifier,
)
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    average_precision_score,
    balanced_accuracy_score,
    brier_score_loss,
    confusion_matrix,
    f1_score,
    precision_recall_curve,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import StratifiedGroupKFold, StratifiedKFold
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler


def make_one_hot_encoder():
    try:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=True,
        )
    except TypeError:
        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=True,
        )


def make_preprocessor(feature_list=None, scale_numeric=False, dense=False):
    feature_list = MODEL_FEATURES if feature_list is None else feature_list
    cat_features, num_features = split_features(feature_list)

    numeric_steps = [
        (
            "impute",
            SimpleImputer(
                strategy="median",
                add_indicator=True,
            ),
        )
    ]

    if scale_numeric:
        numeric_steps.append(
            (
                "scale",
                StandardScaler(with_mean=False),
            )
        )

    categorical_pipeline = Pipeline([
        (
            "impute",
            SimpleImputer(
                strategy="constant",
                fill_value="Missing",
            ),
        ),
        (
            "encode",
            make_one_hot_encoder(),
        ),
    ])

    return ColumnTransformer(
        [
            (
                "numeric",
                Pipeline(numeric_steps),
                num_features,
            ),
            (
                "categorical",
                categorical_pipeline,
                cat_features,
            ),
        ],
        # Histogram gradient boosting needs a dense matrix
        sparse_threshold=0 if dense else 0.3,
    )


def build_majority_baseline(parameters, feature_list=None):
    return Pipeline([
        ("model", DummyClassifier(strategy="prior")),
    ])


def build_logistic_regression(parameters, feature_list=None):
    return Pipeline([
        (
            "preprocess",
            make_preprocessor(feature_list, scale_numeric=True),
        ),
        (
            "model",
            LogisticRegression(max_iter=5000, **parameters),
        ),
    ])


def build_random_forest(parameters, feature_list=None):
    return Pipeline([
        (
            "preprocess",
            make_preprocessor(feature_list, scale_numeric=False),
        ),
        (
            "model",
            RandomForestClassifier(
                n_estimators=300,
                n_jobs=-1,
                random_state=RANDOM_STATE,
                **parameters,
            ),
        ),
    ])


def build_hist_gradient_boosting(parameters, feature_list=None):
    return Pipeline([
        (
            "preprocess",
            make_preprocessor(feature_list, scale_numeric=False, dense=True),
        ),
        (
            "model",
            HistGradientBoostingClassifier(
                early_stopping=False,
                random_state=RANDOM_STATE,
                **parameters,
            ),
        ),
    ])


def build_xgboost(parameters, feature_list=None):
    model = XGBClassifier(
        objective="binary:logistic",
        eval_metric="logloss",
        tree_method="hist",
        n_jobs=-1,
        random_state=RANDOM_STATE,
        **parameters,
    )

    return Pipeline([
        (
            "preprocess",
            make_preprocessor(feature_list, scale_numeric=False),
        ),
        ("model", model),
    ])


def build_neural_network(parameters, feature_list=None):
    model = MLPClassifier(
        activation="relu",
        solver="adam",
        max_iter=300,
        early_stopping=False,
        random_state=RANDOM_STATE,
        **parameters,
    )

    return Pipeline([
        (
            "preprocess",
            make_preprocessor(feature_list, scale_numeric=True),
        ),
        ("model", model),
    ])


def build_catboost(parameters):
    return CatBoostClassifier(
        loss_function="Logloss",
        eval_metric="AUC",
        verbose=False,
        allow_writing_files=False,
        thread_count=-1,
        random_seed=RANDOM_STATE,
        **parameters,
    )


def prepare_catboost_data(frame, feature_list=None):
    feature_list = MODEL_FEATURES if feature_list is None else feature_list
    cat_features, num_features = split_features(feature_list)

    result = frame[feature_list].copy()

    for column in cat_features:
        result[column] = (
            result[column]
            .fillna("Missing")
            .astype(str)
        )

    for column in num_features:
        result[column] = pd.to_numeric(
            result[column],
            errors="coerce",
        )

    result[num_features] = result[
        num_features
    ].replace([np.inf, -np.inf], np.nan)

    return result


PIPELINE_BUILDERS = {
    "Majority baseline": build_majority_baseline,
    "Logistic regression": build_logistic_regression,
    "Random forest": build_random_forest,
    "Histogram gradient boosting": build_hist_gradient_boosting,
    "XGBoost": build_xgboost,
    "Neural network": build_neural_network,
}


def fit_model(model_name, parameters, frame, y, feature_list):
    """Fit any of the seven models on the given rows and feature list."""
    if model_name == "CatBoost":
        cat_features, _ = split_features(feature_list)
        model = build_catboost(parameters)
        model.fit(
            prepare_catboost_data(frame, feature_list),
            y,
            cat_features=cat_features,
        )
    else:
        model = PIPELINE_BUILDERS[model_name](parameters, feature_list)
        model.fit(frame[feature_list], y)
    return model


def predict_model(model, model_name, frame, feature_list):
    if model_name == "CatBoost":
        X = prepare_catboost_data(frame, feature_list)
    else:
        X = frame[feature_list]
    return model.predict_proba(X)[:, 1]


def f1_threshold(y_true, probability):
    """Threshold that maximises F1 (used on development predictions only)."""
    precision_curve, recall_curve, thresholds = precision_recall_curve(
        y_true,
        probability,
    )
    f1_curve = (
        2 * precision_curve * recall_curve
        / np.clip(precision_curve + recall_curve, 1e-12, None)
    )
    return float(thresholds[np.argmax(f1_curve[:-1])])


def calculate_metrics(
    y_true,
    probability,
    threshold=0.5,
):
    prediction = (
        probability >= threshold
    ).astype(int)

    return {
        "ROC-AUC": roc_auc_score(
            y_true,
            probability,
        ),
        "PR-AUC": average_precision_score(
            y_true,
            probability,
        ),
        "Brier": brier_score_loss(
            y_true,
            probability,
        ),
        "Accuracy": accuracy_score(
            y_true,
            prediction,
        ),
        "Balanced accuracy": balanced_accuracy_score(
            y_true,
            prediction,
        ),
        "Precision": precision_score(
            y_true,
            prediction,
            zero_division=0,
        ),
        "Recall": recall_score(
            y_true,
            prediction,
            zero_division=0,
        ),
        "F1": f1_score(
            y_true,
            prediction,
            zero_division=0,
        ),
        "Flagged (%)": prediction.mean() * 100,
    }
```

```python
def cross_validate_configuration(
    model_name,
    parameters,
    feature_list=None,
    folds=None,
):
    """Grouped, stratified 5-fold CV on the development presentations."""
    feature_list = MODEL_FEATURES if feature_list is None else feature_list
    folds = cv_folds if folds is None else folds

    out_of_fold_probability = np.zeros(
        len(development)
    )

    fold_results = []

    for fold_number, (
        train_index,
        validation_index,
    ) in enumerate(folds, start=1):

        model = fit_model(
            model_name,
            parameters,
            development.iloc[train_index],
            y_dev[train_index],
            feature_list,
        )

        probability = predict_model(
            model,
            model_name,
            development.iloc[validation_index],
            feature_list,
        )

        out_of_fold_probability[
            validation_index
        ] = probability

        metrics = calculate_metrics(
            y_dev[validation_index],
            probability,
        )

        metrics["fold"] = fold_number
        fold_results.append(metrics)

    return (
        pd.DataFrame(fold_results),
        out_of_fold_probability,
    )
```

## Student dependency check

Some students have more than one registration: several modules in the same presentation, or a module
repeated or added in a later presentation. Their rows are **not independent**: they share the same person,
habits and circumstances. The cells below measure how strong that dependency is, and show what happens to
cross-validation if it is ignored. This is why every split keeps a student's rows together
(`StratifiedGroupKFold` with `groups = id_student`).

```python
# ---- How many students have more than one registration? ----
registrations_per_student = features.groupby("id_student").size()
presentations_per_student = features.groupby("id_student")["code_presentation"].nunique()
multi_students = registrations_per_student[registrations_per_student > 1].index

same_presentation = features.groupby(["id_student", "code_presentation"]).size()
concurrent_students = (
    same_presentation[same_presentation > 1]
    .index.get_level_values("id_student")
    .unique()
)

dependency_overview = pd.DataFrame({"value": {
    "Registrations": len(features),
    "Students": features["id_student"].nunique(),
    "Students with 1 registration": int((registrations_per_student == 1).sum()),
    "Students with 2 registrations": int((registrations_per_student == 2).sum()),
    "Students with 3 or more registrations": int((registrations_per_student >= 3).sum()),
    "Registrations belonging to students with more than one": int(features["id_student"].isin(multi_students).sum()),
    "Share of all registrations (%)": round(features["id_student"].isin(multi_students).mean() * 100, 1),
    "Students with two modules in the same presentation": len(concurrent_students),
    "Students registered in more than one presentation": int((presentations_per_student > 1).sum()),
}})
display(dependency_overview)

# ---- Are the outcomes of the same student linked? ----
# Every ordered pair of two different registrations of the same student
multi = features.loc[
    features["id_student"].isin(multi_students),
    ["id_student", "module_presentation", "code_presentation", "at_risk"],
]
pairs = multi.merge(multi, on="id_student", suffixes=("_a", "_b"))
pairs = pairs[pairs["module_presentation_a"] != pairs["module_presentation_b"]]
pairs["pair type"] = np.where(
    pairs["code_presentation_a"] == pairs["code_presentation_b"],
    "same presentation (concurrent modules)",
    "different presentations",
)

dependency_rows = []
for pair_type, part in [("all pairs", pairs)] + list(pairs.groupby("pair type")):
    a, b = part["at_risk_a"], part["at_risk_b"]
    dependency_rows.append({
        "pair type": pair_type,
        "pairs": len(part) // 2,
        "P(at risk | other registration at risk)": b[a == 1].mean(),
        "P(at risk | other registration successful)": b[a == 0].mean(),
        "same outcome in both (observed)": (a == b).mean(),
        "same outcome if independent": a.mean() * b.mean() + (1 - a.mean()) * (1 - b.mean()),
        "correlation (phi)": np.corrcoef(a, b)[0, 1],
    })

outcome_dependency = pd.DataFrame(dependency_rows).set_index("pair type")
print("Outcome dependency between registrations of the same student")
display(outcome_dependency.round(3))

# ---- What if the dependency were ignored? Plain vs grouped cross-validation ----
plain_folds = list(
    StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    .split(development, y_dev)
)
plain_overlap = [len(set(groups_dev[tr]) & set(groups_dev[va])) for tr, va in plain_folds]
grouped_overlap = [len(set(groups_dev[tr]) & set(groups_dev[va])) for tr, va in cv_folds]

grouped_metrics, _ = cross_validate_configuration("Logistic regression", {"C": 1.0})
plain_metrics, _ = cross_validate_configuration("Logistic regression", {"C": 1.0}, folds=plain_folds)

display(pd.DataFrame(
    {
        "students in both train and validation (mean per fold)": [
            np.mean(grouped_overlap),
            np.mean(plain_overlap),
        ],
        "CV ROC-AUC": [
            grouped_metrics["ROC-AUC"].mean(),
            plain_metrics["ROC-AUC"].mean(),
        ],
        "CV F1": [
            grouped_metrics["F1"].mean(),
            plain_metrics["F1"].mean(),
        ],
    },
    index=[
        "Grouped by student (used)",
        "Plain stratified (ignores students)",
    ],
).round(4))
print("Grouping keeps every student's rows on one side of each split.")
```

```python
search_rows = []
best_parameters = {}
best_oof_predictions = {}
best_fold_metrics = {}

for model_name, configurations in model_grids.items():
    model_start = time.time()
    best_auc = -np.inf

    print(f"\n{'=' * 60}")
    print(model_name)
    print("=" * 60)

    for configuration_number, parameters in enumerate(
        configurations,
        start=1,
    ):
        configuration_start = time.time()

        fold_metrics, oof_probability = (
            cross_validate_configuration(
                model_name,
                parameters,
            )
        )

        mean_auc = fold_metrics["ROC-AUC"].mean()
        sd_auc = fold_metrics["ROC-AUC"].std()
        mean_pr_auc = fold_metrics["PR-AUC"].mean()
        mean_brier = fold_metrics["Brier"].mean()

        search_rows.append({
            "model": model_name,
            "configuration": configuration_number,
            "parameters": str(parameters),
            "CV ROC-AUC": mean_auc,
            "ROC-AUC SD": sd_auc,
            "CV PR-AUC": mean_pr_auc,
            "CV Brier": mean_brier,
        })

        print(
            f"Configuration {configuration_number}: "
            f"ROC-AUC = {mean_auc:.4f} "
            f"± {sd_auc:.4f}; "
            f"PR-AUC = {mean_pr_auc:.4f}; "
            f"time = "
            f"{time.time() - configuration_start:.0f}s"
        )

        if mean_auc > best_auc:
            best_auc = mean_auc
            best_parameters[model_name] = parameters
            best_oof_predictions[
                model_name
            ] = oof_probability
            best_fold_metrics[
                model_name
            ] = fold_metrics

    print("\nBest parameters:")
    print(best_parameters[model_name])
    print("Best CV ROC-AUC:", round(best_auc, 4))
    print(
        "Total model time:",
        round(time.time() - model_start),
        "seconds",
    )

search_results = (
    pd.DataFrame(search_rows)
    .sort_values(
        ["CV ROC-AUC", "CV PR-AUC"],
        ascending=False,
    )
    .reset_index(drop=True)
)

display(search_results.round(4))
```

```python
# ============================================================
# 1) ALL MODELS ON THE SAME 75 FEATURES (cross-validation)
# ============================================================
model_names = list(model_grids)

comparison_rows = []

for model_name in model_names:
    fold_metrics = best_fold_metrics[model_name]
    comparison_rows.append({
        "Model": model_name,
        "Features": 0 if model_name == "Majority baseline" else len(MODEL_FEATURES),
        "CV ROC-AUC": fold_metrics["ROC-AUC"].mean(),
        "ROC-AUC SD": fold_metrics["ROC-AUC"].std(),
        "CV PR-AUC": fold_metrics["PR-AUC"].mean(),
        "CV Brier": fold_metrics["Brier"].mean(),
        "CV Accuracy (0.5)": fold_metrics["Accuracy"].mean(),
        "CV F1 (0.5)": fold_metrics["F1"].mean(),
    })

comparison = (
    pd.DataFrame(comparison_rows)
    .sort_values(
        "CV ROC-AUC",
        ascending=False,
    )
    .reset_index(drop=True)
)

print("All models, same 75 features, grouped 5-fold CV on 2013B / 2013J / 2014B")
display(comparison.round(4))

# ============================================================
# 2) EVERY MODEL WITH AND WITHOUT THE SCORE FEATURES (CV)
#    Same tuned settings; only the two score features are removed.
# ============================================================
set_fold_metrics = {}
set_oof_predictions = {}

for model_name in model_names:
    # with score: already computed during tuning
    set_fold_metrics[(model_name, MAIN_SET)] = best_fold_metrics[model_name]
    set_oof_predictions[(model_name, MAIN_SET)] = best_oof_predictions[model_name]

    if model_name == "Majority baseline":
        continue  # the baseline uses no features

    start = time.time()
    fold_metrics, oof_probability = cross_validate_configuration(
        model_name,
        best_parameters[model_name],
        FEATURES_WITHOUT_SCORE,
    )
    set_fold_metrics[(model_name, NO_SCORE_SET)] = fold_metrics
    set_oof_predictions[(model_name, NO_SCORE_SET)] = oof_probability
    print(f"{model_name:28s} without score: {time.time() - start:5.0f}s")

score_rows = []
for model_name in model_names:
    if model_name == "Majority baseline":
        continue
    with_score = set_fold_metrics[(model_name, MAIN_SET)]
    without_score = set_fold_metrics[(model_name, NO_SCORE_SET)]
    score_rows.append({
        "Model": model_name,
        "ROC-AUC with score": with_score["ROC-AUC"].mean(),
        "ROC-AUC without score": without_score["ROC-AUC"].mean(),
        "ROC-AUC change": without_score["ROC-AUC"].mean() - with_score["ROC-AUC"].mean(),
        "PR-AUC with score": with_score["PR-AUC"].mean(),
        "PR-AUC without score": without_score["PR-AUC"].mean(),
        "PR-AUC change": without_score["PR-AUC"].mean() - with_score["PR-AUC"].mean(),
        "Brier change": without_score["Brier"].mean() - with_score["Brier"].mean(),
    })

score_check_cv = pd.DataFrame(score_rows).set_index("Model")
print("\nCross-validation: every model with and without the score features")
display(score_check_cv.round(4))

# Decision rule: the score "affects performance" if removing it lowers
# CV ROC-AUC by more than 0.005 (the same margin used for model choice).
SCORE_MARGIN = 0.005
models_hurt = (score_check_cv["ROC-AUC change"] < -SCORE_MARGIN).sum()
print(
    f"Removing the score lowers CV ROC-AUC by more than {SCORE_MARGIN} "
    f"for {models_hurt} of {len(score_check_cv)} models "
    f"(average change {score_check_cv['ROC-AUC change'].mean():+.4f})."
)
print(
    "-> The score features affect performance: keep them."
    if models_hurt > len(score_check_cv) / 2
    else "-> The score features do not affect performance: they can be removed."
)

# ============================================================
# 3) FINAL MODEL AND THRESHOLD (development data only)
# ============================================================
# The group's choice: CatBoost, for the best overall balance between
# finding at-risk students (recall) and avoiding unnecessary alerts
# (precision, F1, Brier). Alternatives depending on university priorities:
#   - "Histogram gradient boosting": find as many at-risk students as possible
#   - "Logistic regression": most explainable, fewest unnecessary alerts
FINAL_MODEL = "CatBoost"

selected_model_name = FINAL_MODEL
selected_parameters = best_parameters[selected_model_name]

# Each model's threshold maximises F1 on its own out-of-fold predictions
model_thresholds = {
    key: (0.5 if key[0] == "Majority baseline" else f1_threshold(y_dev, oof))
    for key, oof in set_oof_predictions.items()
}

selected_oof = set_oof_predictions[(selected_model_name, MAIN_SET)]
SELECTED_THRESHOLD = model_thresholds[(selected_model_name, MAIN_SET)]

best_cv_model = comparison.loc[0, "Model"]
gap = (
    comparison.set_index("Model").loc[best_cv_model, "CV ROC-AUC"]
    - comparison.set_index("Model").loc[selected_model_name, "CV ROC-AUC"]
)

print("Selected model:", selected_model_name)
print("Parameters:", selected_parameters)
print(f"Gap to the best CV model ({best_cv_model}): {gap:.4f} ROC-AUC")
print(
    "Development-only threshold:",
    round(SELECTED_THRESHOLD, 4),
)

development_metrics = calculate_metrics(
    y_dev,
    selected_oof,
    SELECTED_THRESHOLD,
)

display(
    pd.DataFrame(
        development_metrics,
        index=["Development OOF"],
    ).round(4)
)
```

```python
# ============================================================
# FINAL 2014J BENCHMARK
# Models, parameters, features and thresholds are now fixed.
# Every model is refitted on all development registrations
# (same 75 features) and scored once on 2014J.
# Do not tune anything after viewing this result.
# ============================================================

# Retrieve the held-out labels only now.
benchmark_labels = (
    features.loc[
        is_test,
        registration_keys + ["at_risk"],
    ]
    .copy()
    .reset_index(drop=True)
)

benchmark = benchmark_X.merge(
    benchmark_labels,
    on=registration_keys,
    how="inner",
    validate="one_to_one",
)

assert len(benchmark) == len(benchmark_X)
assert benchmark["at_risk"].notna().all()

y_test = benchmark["at_risk"].astype(int).to_numpy()

final_models = {}
test_probabilities = {}
benchmark_rows = []

for model_name in model_names:
    for set_name, feature_list in FEATURE_SETS.items():
        if model_name == "Majority baseline" and set_name != MAIN_SET:
            continue

        model = fit_model(
            model_name,
            best_parameters[model_name],
            development,
            y_dev,
            feature_list,
        )
        probability = predict_model(model, model_name, benchmark, feature_list)

        final_models[(model_name, set_name)] = model
        test_probabilities[(model_name, set_name)] = probability

        threshold = model_thresholds[(model_name, set_name)]
        benchmark_rows.append({
            "Model": model_name,
            "Feature set": set_name,
            "Features": 0 if model_name == "Majority baseline" else len(feature_list),
            "Threshold": threshold,
            **calculate_metrics(y_test, probability, threshold),
        })

benchmark_results = (
    pd.DataFrame(benchmark_rows)
    .set_index(["Model", "Feature set"])
)

print("2014J registrations:", len(y_test))
print("2014J at-risk prevalence:", round(y_test.mean(), 4))

print("\nAll models on the same 75 features (2014J)")
display(
    benchmark_results
    .xs(MAIN_SET, level="Feature set")
    .sort_values("ROC-AUC", ascending=False)
    .round(4)
)

# ---- With vs without the score features on 2014J ----
test_score_rows = []
for model_name in model_names:
    if model_name == "Majority baseline":
        continue
    with_score = benchmark_results.loc[(model_name, MAIN_SET)]
    without_score = benchmark_results.loc[(model_name, NO_SCORE_SET)]
    test_score_rows.append({
        "Model": model_name,
        "ROC-AUC with score": with_score["ROC-AUC"],
        "ROC-AUC without score": without_score["ROC-AUC"],
        "ROC-AUC change": without_score["ROC-AUC"] - with_score["ROC-AUC"],
        "PR-AUC change": without_score["PR-AUC"] - with_score["PR-AUC"],
        "Accuracy change": without_score["Accuracy"] - with_score["Accuracy"],
        "Precision change": without_score["Precision"] - with_score["Precision"],
        "F1 change": without_score["F1"] - with_score["F1"],
        "Flagged (%) with score": with_score["Flagged (%)"],
        "Flagged (%) without score": without_score["Flagged (%)"],
    })

score_check_test = pd.DataFrame(test_score_rows).set_index("Model")
print("\n2014J: every model with and without the score features")
display(score_check_test.round(4))

print("\nAll models without the score features (2014J)")
display(
    benchmark_results
    .xs(NO_SCORE_SET, level="Feature set")
    .sort_values("ROC-AUC", ascending=False)
    .round(4)
)

# ---- Selected model in detail ----
test_probability = test_probabilities[(selected_model_name, MAIN_SET)]

test_prediction = (
    test_probability >= SELECTED_THRESHOLD
).astype(int)

test_metrics = calculate_metrics(
    y_test,
    test_probability,
    SELECTED_THRESHOLD,
)

print("\nFinal model:", selected_model_name)
print("Parameters:", selected_parameters)
print("Fixed threshold:", round(SELECTED_THRESHOLD, 4))

display(
    pd.DataFrame(
        test_metrics,
        index=[f"{selected_model_name} (75 features) - 2014J"],
    ).round(4)
)

confusion = confusion_matrix(
    y_test,
    test_prediction,
)

confusion_table = pd.DataFrame(
    confusion,
    index=[
        "Actually successful",
        "Actually at risk",
    ],
    columns=[
        "Predicted successful",
        "Flagged at risk",
    ],
)

display(confusion_table)
```
